```python
%matplotlib inline 
#%matplotlib inline 是一个常用的魔法命令，用于将Matplotlib绘制的图形直接嵌入到Notebook的输出区域中，而不是弹出单独的窗口。
import matplotlib.pyplot as plt 

x=[1,2,3]
y=[2,4,6]

plt.plot(x,y,marker="o",color="blue")
plt.show()
```

```python
fig,ax=plt.subplots( figsize=(10,7))
ax.plot(x,y,label="data")

ax.set_title("title")
ax.set_xlabel("X")
ax.set_ylabel("Y")

ax.scatter(x,y,label="real")
ax.legend()
plt.show()
```
