```python
import numpy as np 
import tensorflow as tf 

from tensorflow.keras import Sequential 
from tensorflow.keras.layers import Dense 

# 1. 准备训练数据
# 负数标签是 0，正数标签是 1
x_train = np.array([
    [-10.0],
    [-5.0],
    [-2.0],
    [-1.0],
    [1.0],
    [2.0],
    [5.0],
    [10.0]
])

y_train = np.array([
    [0],
    [0],
    [0],
    [0],
    [1],
    [1],
    [1],
    [1]
])


model=Sequential(
    Dense(units=1,activation="sigmoid",input_shape=(1,))
)
model.compile(optimizer="sgd",loss="mean_squared_error",)
model.fit(
    x_train,
    y_train, 
    epochs=200,
    verbose=0,
)

# 5. 测试模型
test_data = np.array([
    [-8.0],
    [-0.5],
    [0.5],
    [8.0]
])

probabilities=model.predit(test_data)
for number, probability in zip(test_data, probabilities):
    value = number[0]
    positive_probability = probability[0]

    if positive_probability >= 0.5:
        result = "正数"
    else:
        result = "负数"

    print(
        f"输入：{value:>5}，"
        f"正数概率：{positive_probability:.4f}，"
        f"判断：{result}"
    )
```
